```python

# ============================================================
# SPOTIFY SONGS' GENRE SEGMENTATION
# Google Colab Complete Project
# ============================================================

import os
import warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import joblib

from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.decomposition import PCA
from sklearn.metrics import silhouette_score
from sklearn.neighbors import NearestNeighbors

warnings.filterwarnings("ignore")
sns.set_theme(style="whitegrid")

RANDOM_STATE = 42
OUTPUT_DIR = "/content/spotify_outputs"
os.makedirs(OUTPUT_DIR, exist_ok=True)

# ============================================================
# 1. LOAD DATASET
# ============================================================

# Upload your CSV in Step 3 before running this cell.
df = pd.read_csv(DATA_PATH)

print("Dataset loaded successfully!")
print("Dataset shape:", df.shape)
print("\nColumn names:", df.columns.tolist())
display(df.head())

# ============================================================
# 2. DATA PREPROCESSING
# ============================================================

print("Duplicate rows:", df.duplicated().sum())
print("\nMissing values:")
display(df.isnull().sum().sort_values(ascending=False).to_frame("Missing"))

df = df.drop_duplicates().copy()

text_columns = [
    "track_name", "track_artist", "playlist_name",
    "playlist_genre", "playlist_subgenre"
]

for col in text_columns:
    if col in df.columns:
        df[col] = df[col].fillna("Unknown").astype(str).str.strip()

numeric_columns = [
    "track_popularity", "danceability", "energy", "key",
    "loudness", "mode", "speechiness", "acousticness",
    "instrumentalness", "liveness", "valence", "tempo",
    "duration_ms"
]

for col in numeric_columns:
    if col in df.columns:
        df[col] = pd.to_numeric(df[col], errors="coerce")

if "track_album_release_date" in df.columns:
    df["track_album_release_date"] = pd.to_datetime(
        df["track_album_release_date"], errors="coerce"
    )
    df["release_year"] = df["track_album_release_date"].dt.year

if "duration_ms" in df.columns:
    df["duration_min"] = df["duration_ms"] / 60000

audio_features = [
    c for c in [
        "danceability", "energy", "loudness", "speechiness",
        "acousticness", "instrumentalness", "liveness",
        "valence", "tempo", "duration_ms"
    ] if c in df.columns
]

if len(audio_features) < 3:
    raise ValueError("Expected Spotify audio feature columns were not found.")

df = df.dropna(subset=audio_features, how="all").reset_index(drop=True)

print("\nCleaned dataset shape:", df.shape)
display(df[audio_features].describe().T)

# ============================================================
# 3. EXPLORATORY DATA ANALYSIS AND VISUALIZATIONS
# ============================================================

# 3.1 Genre distribution
if "playlist_genre" in df.columns:
    plt.figure(figsize=(10, 5))
    sns.countplot(
        data=df, x="playlist_genre",
        order=df["playlist_genre"].value_counts().index
    )
    plt.title("Number of Tracks by Playlist Genre")
    plt.xticks(rotation=35, ha="right")
    plt.tight_layout()
    plt.savefig(f"{OUTPUT_DIR}/01_genre_distribution.png", dpi=150)
    plt.show()

# 3.2 Subgenre distribution
if "playlist_subgenre" in df.columns:
    plt.figure(figsize=(12, 6))
    df["playlist_subgenre"].value_counts().head(15).sort_values().plot(
        kind="barh"
    )
    plt.title("Top 15 Playlist Subgenres")
    plt.xlabel("Number of Rows")
    plt.tight_layout()
    plt.savefig(f"{OUTPUT_DIR}/02_subgenre_distribution.png", dpi=150)
    plt.show()

# 3.3 Top playlist names
if "playlist_name" in df.columns:
    plt.figure(figsize=(12, 7))
    df["playlist_name"].value_counts().head(20).sort_values().plot(
        kind="barh"
    )
    plt.title("Top 20 Playlists by Number of Rows")
    plt.tight_layout()
    plt.savefig(f"{OUTPUT_DIR}/03_top_playlists.png", dpi=150)
    plt.show()

# 3.4 Popularity distribution
if "track_popularity" in df.columns:
    plt.figure(figsize=(9, 5))
    sns.histplot(df["track_popularity"].dropna(), bins=30, kde=True)
    plt.title("Distribution of Track Popularity")
    plt.tight_layout()
    plt.savefig(f"{OUTPUT_DIR}/04_popularity_distribution.png", dpi=150)
    plt.show()

# 3.5 Audio feature distributions
dist_features = [
    c for c in [
        "danceability", "energy", "loudness", "speechiness",
        "acousticness", "instrumentalness", "liveness",
        "valence", "tempo", "duration_min"
    ] if c in df.columns
]

df[dist_features].hist(figsize=(15, 11), bins=30, edgecolor="black")
plt.suptitle("Distribution of Spotify Audio Features", y=1.02)
plt.tight_layout()
plt.savefig(f"{OUTPUT_DIR}/05_audio_feature_distributions.png", dpi=150)
plt.show()

# 3.6 Correlation matrix
corr_features = [
    c for c in audio_features + ["track_popularity"] if c in df.columns
]
corr_matrix = df[corr_features].corr(numeric_only=True)

plt.figure(figsize=(12, 9))
sns.heatmap(
    corr_matrix, annot=True, fmt=".2f",
    cmap="coolwarm", center=0, square=True
)
plt.title("Correlation Matrix of Spotify Features")
plt.tight_layout()
plt.savefig(f"{OUTPUT_DIR}/06_correlation_matrix.png", dpi=150)
plt.show()

display(corr_matrix.round(3))
corr_matrix.to_csv(f"{OUTPUT_DIR}/correlation_matrix.csv")

# 3.7 Average audio features by genre
if "playlist_genre" in df.columns:
    genre_features = [
        c for c in [
            "danceability", "energy", "valence", "acousticness",
            "instrumentalness", "speechiness", "tempo", "loudness"
        ] if c in df.columns
    ]

    genre_means = df.groupby("playlist_genre")[genre_features].mean()

    plt.figure(figsize=(13, 6))
    sns.heatmap(genre_means, annot=True, fmt=".2f", cmap="YlGnBu")
    plt.title("Average Audio Features by Playlist Genre")
    plt.tight_layout()
    plt.savefig(f"{OUTPUT_DIR}/07_genre_audio_heatmap.png", dpi=150)
    plt.show()
    display(genre_means.round(3))

# 3.8 Boxplots of audio features across genres
if "playlist_genre" in df.columns:
    box_features = [
        c for c in [
            "danceability", "energy", "valence",
            "acousticness", "tempo"
        ] if c in df.columns
    ]

    long_df = df.melt(
        id_vars="playlist_genre",
        value_vars=box_features,
        var_name="Feature",
        value_name="Value"
    )

    g = sns.catplot(
        data=long_df, x="playlist_genre", y="Value",
        col="Feature", kind="box", col_wrap=3,
        sharey=False, height=4, aspect=1.2
    )
    g.set_xticklabels(rotation=35)
    g.fig.suptitle("Audio Features Across Genres", y=1.03)
    plt.savefig(
        f"{OUTPUT_DIR}/08_features_by_genre.png",
        dpi=150, bbox_inches="tight"
    )
    plt.show()

# 3.9 Popularity by genre
if {"playlist_genre", "track_popularity"}.issubset(df.columns):
    plt.figure(figsize=(11, 5))
    sns.boxplot(data=df, x="playlist_genre", y="track_popularity")
    plt.title("Track Popularity by Playlist Genre")
    plt.xticks(rotation=35, ha="right")
    plt.tight_layout()
    plt.savefig(f"{OUTPUT_DIR}/09_popularity_by_genre.png", dpi=150)
    plt.show()

# 3.10 Release-year trend
if {"release_year", "track_popularity"}.issubset(df.columns):
    year_data = df.dropna(subset=["release_year"]).copy()
    year_data = year_data[year_data["release_year"] >= 1950]

    if not year_data.empty:
        yearly_popularity = year_data.groupby(
            "release_year"
        )["track_popularity"].mean()

        plt.figure(figsize=(12, 5))
        yearly_popularity.plot()
        plt.title("Average Popularity by Release Year")
        plt.xlabel("Release Year")
        plt.ylabel("Average Popularity")
        plt.tight_layout()
        plt.savefig(f"{OUTPUT_DIR}/10_release_year_trend.png", dpi=150)
        plt.show()

# ============================================================
# 4. PREPARE FEATURES FOR CLUSTERING
# ============================================================

cluster_features = [
    c for c in [
        "danceability", "energy", "loudness", "speechiness",
        "acousticness", "instrumentalness", "liveness",
        "valence", "tempo", "duration_ms"
    ] if c in df.columns
]

imputer = SimpleImputer(strategy="median")
X_imputed = imputer.fit_transform(df[cluster_features])

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X_imputed)

print("Features used:", cluster_features)
print("Feature matrix shape:", X_scaled.shape)

# ============================================================
# 5. CHOOSE NUMBER OF CLUSTERS
# ============================================================

rng = np.random.RandomState(RANDOM_STATE)
sample_size = min(5000, len(X_scaled))
sample_indices = rng.choice(
    len(X_scaled), size=sample_size, replace=False
)
X_sample = X_scaled[sample_indices]

k_values = list(range(2, min(9, len(X_scaled) - 1) + 1))
inertias = []
silhouette_scores = []

for k in k_values:
    model = KMeans(
        n_clusters=k, random_state=RANDOM_STATE, n_init=10
    )
    labels_sample = model.fit_predict(X_sample)
    inertias.append(model.inertia_)
    silhouette_scores.append(
        silhouette_score(X_sample, labels_sample)
    )

fig, axes = plt.subplots(1, 2, figsize=(13, 5))

axes[0].plot(k_values, inertias, marker="o")
axes[0].set_title("Elbow Method")
axes[0].set_xlabel("Number of Clusters (K)")
axes[0].set_ylabel("Inertia")

axes[1].plot(k_values, silhouette_scores, marker="o")
axes[1].set_title("Silhouette Score")
axes[1].set_xlabel("Number of Clusters (K)")
axes[1].set_ylabel("Silhouette Score")

plt.tight_layout()
plt.savefig(f"{OUTPUT_DIR}/11_cluster_selection.png", dpi=150)
plt.show()

best_k = k_values[int(np.argmax(silhouette_scores))]
print("Selected K:", best_k)

display(pd.DataFrame({
    "K": k_values,
    "Inertia": inertias,
    "Silhouette Score": silhouette_scores
}))

# ============================================================
# 6. TRAIN FINAL K-MEANS MODEL
# ============================================================

kmeans = KMeans(
    n_clusters=best_k, random_state=RANDOM_STATE, n_init=10
)
df["cluster"] = kmeans.fit_predict(X_scaled)

final_silhouette = silhouette_score(
    X_sample, df["cluster"].to_numpy()[sample_indices]
)

print("Number of clusters:", best_k)
print("Sample silhouette score:", round(final_silhouette, 4))
display(df["cluster"].value_counts().sort_index().to_frame("Track Rows"))

# ============================================================
# 7. PCA CLUSTER VISUALIZATION
# ============================================================

pca = PCA(n_components=2, random_state=RANDOM_STATE)
X_pca = pca.fit_transform(X_scaled)

pca_df = pd.DataFrame({
    "PC1": X_pca[:, 0],
    "PC2": X_pca[:, 1],
    "cluster": df["cluster"].astype(str).values
})

pca_plot = pca_df.sample(
    min(8000, len(pca_df)), random_state=RANDOM_STATE
)

plt.figure(figsize=(11, 7))
sns.scatterplot(
    data=pca_plot, x="PC1", y="PC2",
    hue="cluster", palette="tab10", alpha=0.6, s=20
)
plt.title("Spotify Song Clusters Using PCA")
plt.tight_layout()
plt.savefig(f"{OUTPUT_DIR}/12_pca_clusters.png", dpi=150)
plt.show()

print(
    "Variance explained by two PCA components:",
    round(pca.explained_variance_ratio_.sum() * 100, 2), "%"
)

# ============================================================
# 8. CLUSTERS BY PLAYLIST GENRE
# ============================================================

if "playlist_genre" in df.columns:
    genre_cluster_counts = pd.crosstab(
        df["cluster"], df["playlist_genre"]
    )
    display(genre_cluster_counts)

    plt.figure(figsize=(12, 6))
    sns.heatmap(
        genre_cluster_counts, annot=True, fmt="d", cmap="Blues"
    )
    plt.title("Track Counts by Cluster and Playlist Genre")
    plt.tight_layout()
    plt.savefig(f"{OUTPUT_DIR}/13_cluster_by_genre_counts.png", dpi=150)
    plt.show()

    genre_cluster_percent = pd.crosstab(
        df["cluster"], df["playlist_genre"], normalize="index"
    ) * 100

    plt.figure(figsize=(12, 6))
    sns.heatmap(
        genre_cluster_percent, annot=True, fmt=".1f", cmap="YlGnBu"
    )
    plt.title("Genre Composition Within Each Cluster (%)")
    plt.tight_layout()
    plt.savefig(f"{OUTPUT_DIR}/14_cluster_by_genre_percent.png", dpi=150)
    plt.show()

    display(genre_cluster_percent.round(2))

# ============================================================
# 9. CLUSTERS BY PLAYLIST NAME
# ============================================================

if "playlist_name" in df.columns:
    playlist_cluster_counts = pd.crosstab(
        df["cluster"], df["playlist_name"]
    )
    print("Playlist distribution across clusters:")
    display(playlist_cluster_counts)

    top_playlists = (
        df.groupby(["cluster", "playlist_name"])
        .size().reset_index(name="track_rows")
        .sort_values(
            ["cluster", "track_rows"],
            ascending=[True, False]
        )
        .groupby("cluster").head(5)
    )

    print("Top 5 playlists in each cluster:")
    display(top_playlists)
    top_playlists.to_csv(
        f"{OUTPUT_DIR}/top_playlists_per_cluster.csv", index=False
    )

# ============================================================
# 10. CLUSTER FEATURE PROFILES
# ============================================================

cluster_profiles = df.groupby("cluster")[cluster_features].mean()

print("Mean audio features in each cluster:")
display(cluster_profiles.round(3))

plt.figure(figsize=(13, 6))
sns.heatmap(
    cluster_profiles, annot=True, fmt=".2f", cmap="viridis"
)
plt.title("Audio Feature Profiles of Each Cluster")
plt.tight_layout()
plt.savefig(f"{OUTPUT_DIR}/15_cluster_feature_profiles.png", dpi=150)
plt.show()

cluster_profiles.to_csv(
    f"{OUTPUT_DIR}/cluster_feature_profiles.csv"
)

# ============================================================
# 11. SONG RECOMMENDATION SYSTEM
# ============================================================

nearest_neighbors = NearestNeighbors(
    metric="cosine", algorithm="brute"
)
nearest_neighbors.fit(X_scaled)


def recommend_songs(track_name, n_recommendations=10, artist=None):
    """Recommend songs using audio-feature similarity."""

    mask = df["track_name"].str.casefold().eq(
        str(track_name).strip().casefold()
    )

    if artist is not None and "track_artist" in df.columns:
        mask &= df["track_artist"].str.casefold().eq(
            str(artist).strip().casefold()
        )

    matches = np.flatnonzero(mask.to_numpy())

    if len(matches) == 0:
        raise ValueError(
            f"Song '{track_name}' not found. "
            "Use an exact song name from the dataset."
        )

    query_idx = matches[0]
    n_search = min(n_recommendations + 1, len(df))

    distances, indices = nearest_neighbors.kneighbors(
        X_scaled[query_idx].reshape(1, -1),
        n_neighbors=n_search
    )

    results = []

    for distance, idx in zip(distances[0], indices[0]):
        if idx == query_idx:
            continue

        row = df.iloc[idx]

        results.append({
            "Track Name": row.get("track_name", "Unknown"),
            "Artist": row.get("track_artist", "Unknown"),
            "Playlist Genre": row.get("playlist_genre", "Unknown"),
            "Playlist Name": row.get("playlist_name", "Unknown"),
            "Cluster": int(row["cluster"]),
            "Similarity Score": round(float(1 - distance), 4)
        })

        if len(results) >= n_recommendations:
            break

    return pd.DataFrame(results)


# Demonstrate recommendations for a real song in the dataset
if "track_name" in df.columns:
    valid_tracks = df.loc[
        df["track_name"].notna() & df["track_name"].ne("Unknown"),
        "track_name"
    ]

    if not valid_tracks.empty:
        example_track = valid_tracks.iloc[0]
        print("Example song:", example_track)
        display(recommend_songs(example_track, 10))

# ============================================================
# 12. SAVE MODEL AND RESULTS
# ============================================================

joblib.dump({
    "cluster_features": cluster_features,
    "imputer": imputer,
    "scaler": scaler,
    "kmeans": kmeans,
    "pca": pca,
    "nearest_neighbors": nearest_neighbors
}, f"{OUTPUT_DIR}/spotify_recommendation_model.joblib")

export_columns = [
    c for c in [
        "track_id", "track_name", "track_artist",
        "track_popularity", "playlist_name",
        "playlist_genre", "playlist_subgenre", "cluster"
    ] if c in df.columns
]

df[export_columns + cluster_features].to_csv(
    f"{OUTPUT_DIR}/spotify_tracks_with_clusters.csv",
    index=False
)

pd.DataFrame({
    "K": k_values,
    "Inertia": inertias,
    "Silhouette Score": silhouette_scores
}).to_csv(f"{OUTPUT_DIR}/cluster_evaluation_scores.csv", index=False)

print("\nPROJECT COMPLETED SUCCESSFULLY!")
print("Rows analyzed:", len(df))
print("Selected clusters:", best_k)
print("Silhouette score:", round(final_silhouette, 4))
print("All output files saved in:", OUTPUT_DIR)
```
